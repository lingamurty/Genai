##Type Dict

```python
# normal dictionary
student_info = {
    'name': 'qt',
    'age': 17,
    'email': 'qt@gmail.com'
}
```

```python
from typing import TypedDict

class StudentInfo(TypedDict):
    name: str
    age: int
    email: str
```

```python
x= StudentInfo(name='qt', age=17, email='qt@gmail.com')
```
