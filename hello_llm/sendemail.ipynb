```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.tools import tool
from langchain.agents import create_agent
```

```python
model=ChatGoogleGenerativeAI(
    model= "gemini-3.5-flash-lite"
)
```

```python
result=model.invoke("what is capital of india")
print(result)
```

```python
@tool
def send_email(content:str,destination:str) -> str:
    """this is for sending mail"""
    print(f'sending email to {destination} content: {content}' )
    print(f'email sent successfully to {destination}')
    return f'email sent successfully to {destination}'
```

```python
agent=create_agent(
    model,
    tools=[send_email],
    system_prompt="you are helpful assistance to access tool")
```

```python
response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "send how are you message email to test@gmail.com"
        }
    ]
})
print(response)
```
