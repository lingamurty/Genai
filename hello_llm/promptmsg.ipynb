```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm=ChatGoogleGenerativeAI(
    model= "gemini-3.5-flash-lite"
)
```

```python
from langchain.tools import tool

@tool
def add(a, b):
    """Add two numbers.

    Args:
        a: The first number.
        b: The second number.

    Returns:
        The sum of a and b.
    """
    return a + b

@tool
def sub(a, b):
    """Subtract the second number from the first number.

    Args:
        a: The number to subtract from.
        b: The number to subtract.

    Returns:
        The difference between a and b.
    """
    return a - b

@tool
def multiply(a, b):
    """Multiply two numbers.

    Args:
        a: The first number.
        b: The second number.

    Returns:
        The product of a and b.
    """
    return a * b

@tool
def divide(a, b):
    """Divide the first number by the second number.

    Args:
        a: The numerator.
        b: The denominator.

    Returns:
        The quotient of a divided by b.

    Raises:
        ZeroDivisionError: If b is zero.
    """
    return a / b

@tool
def modulus(a, b):
    """Calculate the remainder after dividing a by b.

    Args:
        a: The dividend.
        b: The divisor.

    Returns:
        The remainder of a divided by b.

    Raises:
        ZeroDivisionError: If b is zero.
    """
    return a % b
```

```python
from langchain.agents import create_agent
agent = create_agent(
    model=llm,
    tools=[add,sub,multiply,modulus]
        
    )
```

```python
def print_agent_response(response):
    for message in response['messages']:
        print(message)
```

```python
agent_without_shortterm_memory = create_agent(
    model='google_genai:gemini-3.5-flash-lite'
)

response = agent_without_shortterm_memory.invoke(
    {
        "messages": "My name is Lingamurty"
    }
)
print_agent_response(response)

response = agent_without_shortterm_memory.invoke(
    {
        "messages": "What is my name?"
    }
)
print_agent_response(response)
```

```python
from langgraph.checkpoint.memory import InMemorySaver
cfg_1 = {"configurable": {"thread_id": "classroom-02sep"}}
cfg_2 = {"configurable": {"thread_id": "klsdjfdaslk"}}

agent_with_shortterm_memory = create_agent(
    model='google_genai:gemini-3.5-flash-lite',
    checkpointer=InMemorySaver(),
)

response = agent_with_shortterm_memory.invoke(
    {
        "messages": "My name is Lingamurty"
    },
    cfg_1
)
print_agent_response(response)
```

```python
# agent with short term same thread/conversation
response = agent_with_shortterm_memory.invoke(
    {
        "messages": "What is my name?"
    },
    cfg_1
)
print_agent_response(response)
```

```python
# agent with short term different thread/conversation
response = agent_with_shortterm_memory.invoke(
    {
        "messages": "What is my name?"
    },
    cfg_2
)
print_agent_response(response)
```

```python
#longterm memory
from langgraph.store.memory import InMemoryStore

store = InMemoryStore()

store.put(('hr', 'hyd', 'abc'),'favorites', {'color': 'blue', 'food': 'biriyani'} )
store.put(('it', 'hyd', 'xyz'),'favorites', {'color': 'red', 'food': 'dosa'} )
```

```python
store.list_namespaces()
```

```python
store.get(('hr', 'hyd', 'abc'), 'favorites')
```

```python
store.search(
    ('hr', 'hyd', 'abc'),
    query='biryani'
)
```
